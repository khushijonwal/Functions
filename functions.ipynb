 6. Explain the concept of generators in Python and how they are defined.
        -Generators are special functions that return values one at a time using yield instead of return.
        Example :
        def countdown(n):
             while n > 0:
                 yield n
                 n -= 1

        for i in countdown(3):
            print(i)

 10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
 
        - map() is used to apply a function to every item in an iterable and returns the transformed result.
        - filter() is used to filter out elements based on a condition. Only the items that return True will be included.
        - reduce() (from functools) applies a rolling computation to reduce the iterable to a single value.
        Example -
            list(map(lambda x: x+1, [1, 2, 3]))  

            list(filter(lambda x: x % 2 == 0, [1, 2, 3]))  

            reduce(lambda x, y: x + y, [1, 2, 3, 4])

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

```python
def sum_even(numbers):
    return sum(n for n in numbers if n % 2 == 0)

print(sum_even([1, 2, 3, 4, 5, 6]))
```

3.  Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

# Example
print(is_prime(13))  # Output: True
print(is_prime(100))
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

```python
class Fibonacci:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value


for num in Fibonacci(7):
    print(num, end=' ')  
```

6. Write a generator function in Python that yields the powers of 2 up to given exponent.

```python
def powers_of_two(max_exp):
    for i in range(max_exp + 1):
        yield 2 ** i

for val in powers_of_two(4):
    print(val, end=' ')  
```

7. Implement a generator function that reads a file line by line and yields each line as a string.

```python
def read_file_lines(filepath):
    with open(filepath, 'r') as f:
        for line in f:
            yield line.strip()


for line in read_file_lines("file.txt"):
   print(line)
```

8.  Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

```python
data = [(1, 5), (3, 1), (2, 9), (4, 0)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)  
```

9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
celsius = [0, 10, 20, 30]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)  
```

10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return ''.join(filter(lambda ch: ch not in vowels, s))

print(remove_vowels("Beautiful Day")) 
```
